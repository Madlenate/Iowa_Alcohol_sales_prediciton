# Next Step (PCA)

### The next step is to understand if the data is usuable in describing the variance or how the sales prices continues to grow.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

cat_df = q("""
    SELECT
        county_name,
        sale_year,
        EXTRACT(month FROM ordered_on) AS month,
        category_name,
        SUM(sales_dollars) AS cat_dollars -- Summ of Dollars 
    FROM sales
    WHERE county_name IS NOT NULL
    GROUP BY 1,2,3,4
""")

main_df = q("""
    SELECT
        county_name,
        sale_year,
        EXTRACT(month FROM ordered_on) AS month,
        SUM(sales_dollars) AS total_dollars,
        SUM(sales_bottles) AS total_bottles,
        SUM(sales_liters) AS total_liters,
        COUNT(DISTINCT invoice_id) AS n_transactions,
        COUNT(DISTINCT store_no) AS n_stores,
        COUNT(DISTINCT vendor_number) AS n_vendors,
        AVG(pack) AS avg_pack,
        AVG(bottle_volume_ml) AS avg_bottle_volume_ml
    FROM sales
    WHERE county_name IS NOT NULL
    GROUP BY 1,2,3
""")

```

```python
population_df = q("""
    SELECT county_name, pop_year AS sale_year, population
    FROM county_population
""")

grp_total = cat_df.groupby(["county_name", "sale_year", "month"])["cat_dollars"].transform(sum)
cat_df["share"] = cat_df["cat_dollars"] / grp_total #probability of county sale and month 
cat_df["p_logp"] = -cat_df["share"] * np.log(cat_df["share"].where(cat_df["share"] > 0)) # helps us caculate the level of uncertainty, randomness, or impurity of a dataset

entropy = (
    cat_df.groupby(["county_name", "sale_year", "month"])["p_logp"]
    .sum()
    .reset_index(name="category_entropy")
)
n_categories = (
    cat_df.groupby(["county_name", "sale_year", "month"])["category_name"]
    .nunique()
    .reset_index(name="n_categories")
)

df = main_df.merge(entropy, on=["county_name", "sale_year", "month"])
df = df.merge(n_categories,  on=["county_name", "sale_year", "month"])
df = df.merge(unemployment_df, on=["sale_year", "month"])  # national monthly rate, same value broadcast to every county that month
df = df.merge(
    homeless_df[["pit_year", "overall_homeless", "overall_chronically_homeless", "unsheltered_homeless"]].rename(columns={"pit_year": "sale_year"}),
    on="sale_year", how="left"  # annual statewide count, same value broadcast to every county-month in that year; left join since homelessness data doesn't yet cover 2026
)
df = df.merge(population_df, on=["county_name", "sale_year"], how="left")  # annual county population, same value broadcast to every month in that year; left join since EL PASO (not a real Iowa county) won't match
df["avg_price_per_bottle"] = df["total_dollars"] / df["total_bottles"]
df["stores_per_10k_capita"] = df["n_stores"] / df["population"] * 10_000  # store density -- normalizes n_stores for county size instead of raw count

print(f"County-month rows: {df.shape[0]}, counties: {df['county_name'].nunique()}")
df.head()
```

```python
feature_cols = [
    "n_stores", "n_vendors", "n_categories",
    "avg_pack", "avg_bottle_volume_ml", "avg_price_per_bottle",
    "category_entropy", "unemployment_rate",
    "overall_homeless", "overall_chronically_homeless", "unsheltered_homeless", "stores_per_10k_capita"
]
X = df[feature_cols].dropna() # X training collumns 

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA()
pcs = pca.fit_transform(X_scaled)

evr = pca.explained_variance_ratio_
cum = np.cumsum(evr)

for i, (e, c) in enumerate(zip(evr, cum), start=1):
    print(f"PC{i}: {e} cumulative: {c}")
```

```python
loadings = pd.DataFrame(
    pca.components_.T * np.sqrt(pca.explained_variance_),
    index=feature_cols,
    columns=[f"PC{i+1}" for i in range(len(feature_cols))],
)
corr = X.corr()

fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(corr.values, cmap="coolwarm", vmin=-1, vmax=1, aspect="auto")

ax.set_xticks(range(corr.shape[1]))
ax.set_xticklabels(corr.columns, rotation=45, ha="right")
ax.set_yticks(range(corr.shape[0]))
ax.set_yticklabels(corr.index)

for i in range(corr.shape[0]):
    for j in range(corr.shape[1]):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center", fontsize=7)

fig.colorbar(im, ax=ax, label="correlation")
ax.set_title("Feature correlation matrix")
plt.tight_layout()
plt.show()



fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(loadings.values, cmap="coolwarm", vmin=-1, vmax=1, aspect="auto")

ax.set_xticks(range(loadings.shape[1]))
ax.set_xticklabels(loadings.columns)
ax.set_yticks(range(loadings.shape[0]))
ax.set_yticklabels(loadings.index)

for i in range(loadings.shape[0]):
    for j in range(loadings.shape[1]):
        ax.text(j, i, f"{loadings.iloc[i, j]:.2f}", ha="center", va="center", fontsize=7)

fig.colorbar(im, ax=ax, label="loading")
ax.set_title("PCA component loadings")
plt.tight_layout()
plt.show()


```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].bar(range(1, len(evr) + 1), evr, color="#4C72B0", label="individual")
axes[0].plot(range(1, len(evr) + 1), cum, color="#C44E52", marker="o", label="cumulative")
axes[0].set_xlabel("Principal component")
axes[0].set_ylabel("Explained variance ratio")
axes[0].set_title("Scree plot")
axes[0].legend()
axes[0].set_ylim(0, 1.05)

axes[1].scatter(pcs[:, 0], pcs[:, 1], s=6, alpha=0.3, color="#4C72B0")
for i, feat in enumerate(feature_cols):
    axes[1].arrow(0, 0, loadings.iloc[i, 0]*3, loadings.iloc[i, 1]*3,
                  color="#C44E52", head_width=0.15, alpha=0.8)
    axes[1].text(loadings.iloc[i, 0]*3.3, loadings.iloc[i, 1]*3.3, feat, fontsize=7, color="#C44E52")
axes[1].set_xlabel("PC1")
axes[1].set_ylabel("PC2")
axes[1].set_title("PC1 vs PC2 (county-months) with feature loadings")
axes[1].axhline(0, color="grey", lw=0.5)
axes[1].axvline(0, color="grey", lw=0.5)

plt.tight_layout()
plt.show()
```

# Applying the linear regression line

```python
statewide_full = (
    df[df["sale_year"] < 2026]  # 2026 is still in progress -- a partial year would understate its total and distort the trend
    .groupby(["sale_year", "month"])
    .agg(
        total_dollars=("total_dollars", "sum"),
        total_bottles=("total_bottles", "sum"),
        n_stores=("n_stores", "sum"),  # sum of per-county distinct store counts & approximates statewide active stores that month
        n_vendors=("n_vendors", "sum"),  # sum of per-county distinct vendor (brand) counts & approximates statewide active vendors that month
        population=("population", "sum"),  # sum of matched counties' population that year (excludes EL PASO, which has no match)
        unemployment_rate=("unemployment_rate", "mean"),  # same national value for every county that month, so mean == the value
        overall_homeless=("overall_homeless", "mean"),  # annual statewide count, same value for every month in a year, so mean == the value
        overall_chronically_homeless=("overall_chronically_homeless", "mean"),
        unsheltered_homeless=("unsheltered_homeless", "mean"),
    )
    .reset_index()
)
statewide_full["avg_price_per_bottle"] = statewide_full["total_dollars"] / statewide_full["total_bottles"]
statewide_full["stores_per_10k_capita"] = statewide_full["n_stores"] / statewide_full["population"] * 10_000  # recomputed at the aggregate level, not averaged from per-county ratios
statewide_full["time_index"] = (statewide_full["sale_year"] - statewide_full["sale_year"].min()) * 12 + statewide_full["month"]
statewide_full["log_dollars"] = np.log1p(statewide_full["total_dollars"])
sf_month_dummies = pd.get_dummies(statewide_full["month"], prefix="month", drop_first=True)

statewide_full.head()

```

```python
monthly_plot_df = statewide_full.sort_values(["sale_year", "month"]).copy()
monthly_plot_df["date"] = pd.to_datetime(dict(year=monthly_plot_df["sale_year"], month=monthly_plot_df["month"], day=1))

indexed_monthly = monthly_plot_df.set_index("date")[
    ["total_dollars", "n_stores", "n_vendors", "avg_price_per_bottle", "unemployment_rate", "overall_homeless", "stores_per_10k_capita"]
]
indexed_monthly = indexed_monthly / indexed_monthly.iloc[0] * 100  # index each series to 100 at the first month

fig, ax = plt.subplots(figsize=(20, 9))
ax.plot(indexed_monthly.index, indexed_monthly["total_dollars"], label="total sales dollars")
ax.plot(indexed_monthly.index, indexed_monthly["n_stores"], label="number of stores")
ax.plot(indexed_monthly.index, indexed_monthly["n_vendors"], label="number of vendors (brands)")
ax.plot(indexed_monthly.index, indexed_monthly["avg_price_per_bottle"], label="avg price per bottle")
ax.plot(indexed_monthly.index, indexed_monthly["unemployment_rate"], label="unemployment rate")
ax.plot(indexed_monthly.index, indexed_monthly["overall_homeless"], label="overall homelessness")
ax.plot(indexed_monthly.index, indexed_monthly["stores_per_10k_capita"], label="stores per 10k capita")
ax.axhline(100, color="grey", lw=0.5)
ax.set_ylabel(f"index (100 = {monthly_plot_df['date'].iloc[0]:%b %Y} level)")
ax.set_xlabel("date")
ax.set_title("Sales, store count, vendors, price, unemployment, homelessness, and store density -- indexed monthly, 2012-2025")
ax.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

def fit_model(cols, label):
    # October 2025 has a blank unemployment_rate in the source FRED data (likely tied to
    # the Oct 2025 government shutdown delaying that month's BLS release) -- drop it only
    # when the model actually uses that column, so other models keep the full sample.
    data = pd.concat([statewide_full[cols + ["log_dollars"]], sf_month_dummies], axis=1).dropna()
    X = data[cols + list(sf_month_dummies.columns)]
    y = data["log_dollars"]
    model = LinearRegression().fit(X, y)
    r2 = r2_score(y, model.predict(X))
    coefs = dict(zip(cols, model.coef_[: len(cols)].round(5)))
    print(f"{label:35s} rows={len(data):3d}  R^2={r2:.3f}  coefs={coefs}")
    return model, r2

model_trend, r2_trend = fit_model(["time_index"], "trend only")
model_drivers, r2_drivers = fit_model(["n_stores", "avg_price_per_bottle"], "n_stores + price")
model_combined, r2_combined = fit_model(["time_index", "n_stores", "avg_price_per_bottle"], "trend + n_stores + price")
model_unemployment, r2_unemployment = fit_model(["unemployment_rate"], "unemployment only")
model_all, r2_all = fit_model(["n_stores", "avg_price_per_bottle", "unemployment_rate"], "n_stores + price + unemployment")
model_homeless, r2_homeless = fit_model(["overall_homeless"], "homelessness only")
model_chronic_homeless, r2_chronic_homeless = fit_model(["overall_chronically_homeless"], "chronic homelessness only")
model_unsheltered_homeless, r2_unsheltered_homeless = fit_model(["unsheltered_homeless"], "unsheltered homelessness only")
model_drivers_homeless, r2_drivers_homeless = fit_model(["n_stores", "avg_price_per_bottle", "overall_homeless"], "n_stores + price + homelessness")
model_drivers_unsheltered, r2_drivers_unsheltered = fit_model(
    ["n_stores", "avg_price_per_bottle", "unsheltered_homeless"], "n_stores + price + unsheltered homelessness"
)
model_everything, r2_everything = fit_model(
    ["n_stores", "avg_price_per_bottle", "unemployment_rate", "overall_homeless", "unsheltered_homeless"], "everything"
)
model_density, r2_density = fit_model(["stores_per_10k_capita"], "store density only")
model_density_price, r2_density_price = fit_model(
    ["stores_per_10k_capita", "avg_price_per_bottle"], "store density + price"
)
model_everything_density, r2_everything_density = fit_model(
    ["stores_per_10k_capita", "avg_price_per_bottle", "unemployment_rate", "overall_homeless", "unsheltered_homeless"],
    "everything (density instead of n_stores)"
)

```

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import Ridge, Lasso, ElasticNet

cv_cols = ["stores_per_10k_capita", "avg_price_per_bottle", "unemployment_rate", "overall_homeless", "unsheltered_homeless"]
cv_data = pd.concat([statewide_full[cv_cols + ["log_dollars"]], sf_month_dummies], axis=1).dropna()
X_cv = cv_data[cv_cols + list(sf_month_dummies.columns)]
y_cv = cv_data["log_dollars"]

regression_models = {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=0.01),
    "ElasticNet": ElasticNet(alpha=0.01, l1_ratio=0.5),
}

kf = KFold(n_splits=5, shuffle=True, random_state=0)
results = []
for name, model in regression_models.items():
    scores = cross_val_score(model, X_cv, y_cv, cv=kf, scoring="r2")
    results.append((name, scores.mean(), scores.std()))

results_df = pd.DataFrame(results, columns=["model", "cv_r2_mean", "cv_r2_std"]).sort_values("cv_r2_mean", ascending=False)
results_df
```

```python
main_df = q("""
    SELECT
        sale_year,
        EXTRACT(month FROM ordered_on) AS month,
        SUM(sales_dollars) AS total_dollars,
        SUM(sales_bottles) AS total_bottles,
        COUNT(DISTINCT store_no) AS n_stores,
        COUNT(DISTINCT vendor_number) AS n_vendors
    FROM sales
    WHERE county_name IS NOT NULL AND sale_year < 2026
    GROUP BY 1, 2
""")

unemployment_df = q("""
    SELECT obs_year AS sale_year, obs_month AS month, unemployment_rate
    FROM unemployment
""")

homeless_df = q("SELECT * FROM homelessness")

population_df = q("""
    SELECT pop_year AS sale_year, SUM(population) AS population
    FROM county_population
    GROUP BY 1
""")

statewide_full = main_df.merge(unemployment_df, on=["sale_year", "month"])
statewide_full = statewide_full.merge(
    homeless_df[["pit_year", "overall_homeless", "overall_chronically_homeless", "unsheltered_homeless"]].rename(columns={"pit_year": "sale_year"}),
    on="sale_year", how="left"
)
statewide_full = statewide_full.merge(population_df, on="sale_year", how="left")
statewide_full = statewide_full.sort_values(["sale_year", "month"]).reset_index(drop=True)  # fix: SQL GROUP BY doesn't guarantee row order, which broke the rolling-window residual calc downstream

statewide_full["avg_price_per_bottle"] = statewide_full["total_dollars"] / statewide_full["total_bottles"]
statewide_full["stores_per_10k_capita"] = statewide_full["n_stores"] / statewide_full["population"] * 10_000
statewide_full["time_index"] = (statewide_full["sale_year"] - statewide_full["sale_year"].min()) * 12 + statewide_full["month"]
statewide_full["log_dollars"] = np.log1p(statewide_full["total_dollars"])
sf_month_dummies = pd.get_dummies(statewide_full["month"], prefix="month", drop_first=True)

statewide_full.head()
```

```python
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.pipeline import make_pipeline


# Step 1: fit the baseline model and get residuals (actual - predicted log_dollars) **EMBEDDING**
baseline_cols = ["n_stores", "avg_price_per_bottle"]
baseline_data = pd.concat([statewide_full[baseline_cols + ["log_dollars"]], sf_month_dummies], axis=1).dropna() #combining everything into one data set
X_base = baseline_data[baseline_cols + list(sf_month_dummies.columns)] # columns along with the dummies data columns for feateres
y_base = baseline_data["log_dollars"] #target variable 
baseline_model = LinearRegression().fit(X_base, y_base) #fitting Linear regression to baseline_model
statewide_full.loc[baseline_data.index, "residual"] = y_base - baseline_model.predict(X_base) # for every index within baseline_data, we get the residuaal of every prediction to the actual within the baseline data

# Step 2: rule-based dip label -- a SUSTAINED downturn, not a one-off bad month.
# Rolling 3-month average residual below -0.5 SD filters out single-month noise.
resid_std = statewide_full["residual"].std() 
statewide_full["rolling_resid"] = statewide_full["residual"].rolling(3, center=True).mean()
statewide_full["dip"] = (statewide_full["rolling_resid"] < -0.5 * resid_std).astype("Int64") #essentially calculating every 3 rolls of residuals

print(f"Dip months flagged: {statewide_full['dip'].sum()} / {statewide_full['dip'].notna().sum()}")
print(statewide_full.loc[statewide_full["dip"] == 1, ["sale_year", "month", "residual", "rolling_resid"]]) # Looks for all dips that are negative

```

```python
policy_df = q("SELECT policy_year AS sale_year, post_2019_spirits_deregulation, post_2023_spirits_expansion FROM alcohol_policy")
statewide_full = statewide_full.merge(policy_df, on="sale_year", how="left") #this involves any years that possibly hahve a policy change that might be note worth weather or not there is a dip or change

clf_cols = ["n_stores", "n_vendors", "avg_price_per_bottle", "unemployment_rate", "overall_homeless", "unsheltered_homeless", "post_2023_spirits_expansion"]
clf_data = pd.concat([statewide_full[clf_cols + ["dip"]], sf_month_dummies], axis=1).dropna() #Combining everything
X_clf = clf_data[clf_cols + list(sf_month_dummies.columns)] 
y_clf = clf_data["dip"].astype(int) # Target for our classification model

log_reg = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, class_weight="balanced"))
log_reg.fit(X_clf, y_clf)
y_pred = log_reg.predict(X_clf)
print(classification_report(y_clf, y_pred, target_names=["normal", "dip"]))
print("Confusion matrix:\n", confusion_matrix(y_clf, y_pred))

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
cv_scores = cross_val_score(log_reg, X_clf, y_clf, cv=skf, scoring="f1")
print(f"\nCross-validated F1 (dip class): {cv_scores.mean()} (+/- {cv_scores.std()})")

coefs = pd.Series(log_reg.named_steps["logisticregression"].coef_[0][:len(clf_cols)], index=clf_cols).sort_values(key=abs, ascending=False)
print("\nStandardized coefficients:")
print(coefs)
```

```python
statewide_full = statewide_full.sort_values(["sale_year", "month"]).reset_index(drop=True)

#Autoregressive: Meaning we want to create something that will allow the machine to momentum to go to the next motnh's residual 
statewide_full["residual_lag1"] = statewide_full["residual"].shift(1) #by shifting the residual by one row down it will teach the other rows 

#Growth rates instead of levels "is this month up or down"
# has predictive power, not the raw store count itself
statewide_full["n_stores_lag1"] = statewide_full["n_stores"].shift(1)
statewide_full["price_lag1"] = statewide_full["avg_price_per_bottle"].shift(1)
statewide_full["n_vendors_lag1"] = statewide_full["n_vendors"].shift(1)
statewide_full["overall_homeless_lag1"] = statewide_full["overall_homeless"].shift(1)
statewide_full["post_2023_lag1"] = statewide_full["post_2023_spirits_expansion"].shift(1)

# Simpler way of getting months instead of using dummies
statewide_full["month_sin"] = np.sin(2 * np.pi * statewide_full["month"] / 12)
statewide_full["month_cos"] = np.cos(2 * np.pi * statewide_full["month"] / 12)

#shifts one down or to teach the model what we wan't to forecast 
statewide_full["log_dollars_next"] = statewide_full["log_dollars"].shift(-1)
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet, HuberRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, WhiteKernel #need to explore more
from xgboost import XGBRegressor
from statsmodels.tsa.statespace.sarimax import SARIMAX # NEED to explore more
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import r2_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler


forecast_cols = ["time_index", "residual_lag1", "n_stores_lag1", "price_lag1"]
seasonal_cols = ["month_sin", "month_cos"]  # UPDATED: replaces sf_month_dummies (11 cols -> 2 cyclical cols)
fc_data = statewide_full[forecast_cols + seasonal_cols + ["log_dollars_next"]].dropna()  


forecast_candidates = {
    "Linear Regression": LinearRegression(),
    "Ridge": make_pipeline(StandardScaler(), Ridge(alpha=1.0)),
    "Lasso": make_pipeline(StandardScaler(), Lasso(alpha=0.01)),
    "ElasticNet": make_pipeline(StandardScaler(), ElasticNet(alpha=0.01, l1_ratio=0.5)),
    "Huber": make_pipeline(StandardScaler(), HuberRegressor(max_iter=1000)),  # also bumped max_iter as a backstop
    "Random Forest": RandomForestRegressor(n_estimators=200, max_depth=4, random_state=0),
    "XGBoost": XGBRegressor(n_estimators=200, max_depth=3, learning_rate=0.05, random_state=0),
    "Gaussian Process": make_pipeline(
        StandardScaler(),
        GaussianProcessRegressor(
            kernel=RBF() + WhiteKernel(noise_level_bounds=(1e-7, 1e5)),  # widened -- default lower bound (1e-5) was too tight for this noisy, small-n data
            normalize_y=True, random_state=0
        )
    )
}
```

```python
tscv = TimeSeriesSplit(n_splits=5)
cv_results = []

forecast_cols = ["time_index", "residual_lag1", "n_stores_lag1", "price_lag1"]
seasonal_cols = ["month_sin", "month_cos"]
fc_data = statewide_full[forecast_cols + seasonal_cols + ["log_dollars_next"]].dropna()

X_forecast = fc_data[forecast_cols + seasonal_cols]
y_forecast = fc_data["log_dollars_next"]

naive_pred_all = statewide_full["log_dollars"].shift(11)
naive_pred_aligned = naive_pred_all.loc[fc_data.index]

naive_fold_scores = []
for train_idx, test_idx in tscv.split(X_forecast):
    y_true = y_forecast.iloc[test_idx]
    y_naive = naive_pred_aligned.iloc[test_idx]
    mask = y_naive.notna()
    if mask.any():
        naive_fold_scores.append(r2_score(y_true[mask], y_naive[mask]))

if naive_fold_scores:
    cv_results.append(("Naive (same month last year)", np.mean(naive_fold_scores), np.std(naive_fold_scores)))



for name, model in forecast_candidates.items():
    fold_scores = []
    for train_idx, test_idx in tscv.split(X_forecast):
        X_train, X_test = X_forecast.iloc[train_idx], X_forecast.iloc[test_idx]
        y_train, y_test = y_forecast.iloc[train_idx], y_forecast.iloc[test_idx]
        model.fit(X_train, y_train)
        fold_scores.append(r2_score(y_test, model.predict(X_test)))
    cv_results.append((name, np.mean(fold_scores), np.std(fold_scores))) #this will go into the data frame later to allow us to crossvalidate 

# NOTE need to study this a little more because I am unfamiliar with this model. Want to understand how it works and how I i can utilize it in the future
sarimax_fold_scores = []
for train_idx, test_idx, in tscv.split(X_forecast):
    y_train, y_test = y_forecast.iloc[train_idx], y_forecast[test_idx]
    exog_train = X_forecast.iloc[train_idx][["n_stores_lag1", "price_lag1"]]
    exog_test = X_forecast.iloc[test_idx][["n_stores_lag1", "price_lag1"]]
    try:
        sarimax_fit = SARIMAX(
            y_train, exog=exog_train, order=(1, 0, 0), seasonal_order=(1, 0, 0, 12),
            enforce_stationarity=False, enforce_invertibility=False
        ).fit(disp=False)
        preds = sarimax_fit.forecast(steps=len(test_idx), exog=exog_test)
        sarimax_fold_scores.append(r2_score(y_test, preds))
    except Exception as e:
        print(f"SARIMAX fold skipped: {e}")

if sarimax_fold_scores:
    cv_results.append(("SARIMAX", np.mean(sarimax_fold_scores), np.std(sarimax_fold_scores)))

cv_results_df = pd.DataFrame(
    cv_results, columns=["model", "cv_r2_mean", "cv_r2_std"]
).sort_values("cv_r2_mean", ascending=False)

cv_results_df
```

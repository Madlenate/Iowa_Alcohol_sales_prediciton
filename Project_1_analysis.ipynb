```python
county_year_stats = q("""
    SELECT county_name, sale_year,
        SUM(sales_dollars) AS total_dollars,
        SUM(sales_liters) AS total_liters, 
        COUNT(DISTINCT store_no) AS n_stores
    FROM sales
    WHERE county_name IS NOT NULL AND sale_year < 2026
    GROUP BY county_name, sale_year
""")

def padded_xlim(values, fmt):
    labels = [fmt.format(v) for v in values]
    return 0, values.max() * (1 + 0.025 * max(len(s) for s in labels))

for year in sorted(county_year_stats["sale_year"].unique()):
    yearly = county_year_stats[county_year_stats["sale_year"] == year]
    top_counties = yearly.nlargest(15, "total_dollars").sort_values("total_dollars")

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

    sales_millions = top_counties["total_dollars"] / 1_000_000
    ax1.barh(top_counties["county_name"], sales_millions, color="#4C72B0")
    ax1.bar_label(ax1.containers[0], fmt="${:,.1f}M", padding=3, fontsize=8)
    ax1.set_xlim(*padded_xlim(sales_millions, "${:,.1f}M"))
    ax1.set_xlabel("sales ($ millions)")
    ax1.set_title(f"Top 15 counties by liquor sales -- {year}")

    ax2.barh(top_counties["county_name"], top_counties["n_stores"], color="#C44E52")
    ax2.bar_label(ax2.containers[0], fmt="{:,.0f}", padding=3, fontsize=8)
    ax2.set_xlim(*padded_xlim(top_counties["n_stores"], "{:,.0f}"))
    ax2.set_xlabel("unique stores")
    ax2.set_title(f"Store count, same counties -- {year}")

    fig.suptitle(f"Iowa liquor sales vs. store count by county -- {year}")
    plt.tight_layout()
    plt.show()
```

# Next Step (PCA)

### The next step is to understand if the data is usuable in describing the variance or how the sales prices continues to grow.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

cat_df = q("""
    SELECT
        county_name,
        sale_year,
        EXTRACT(month FROM ordered_on) AS month,
        category_name,
        SUM(sales_dollars) AS cat_dollars -- Summ of Dollars 
    FROM sales
    WHERE county_name IS NOT NULL
    GROUP BY 1,2,3,4
""")

main_df = q("""
    SELECT
        county_name,
        sale_year,
        EXTRACT(month FROM ordered_on) AS month,
        SUM(sales_dollars) AS total_dollars,
        SUM(sales_bottles) AS total_bottles,
        SUM(sales_liters) AS total_liters,
        COUNT(DISTINCT invoice_id) AS n_transactions,
        COUNT(DISTINCT store_no) AS n_stores,
        COUNT(DISTINCT vendor_number) AS n_vendors,
        AVG(pack) AS avg_pack,
        AVG(bottle_volume_ml) AS avg_bottle_volume_ml
    FROM sales
    WHERE county_name IS NOT NULL
    GROUP BY 1,2,3
""")

```

```python
population_df = q("""
    SELECT county_name, pop_year AS sale_year, population
    FROM county_population
""")

grp_total = cat_df.groupby(["county_name", "sale_year", "month"])["cat_dollars"].transform(sum)
cat_df["share"] = cat_df["cat_dollars"] / grp_total #probability of county sale and month 
cat_df["p_logp"] = -cat_df["share"] * np.log(cat_df["share"].where(cat_df["share"] > 0)) # helps us caculate the level of uncertainty, randomness, or impurity of a dataset

entropy = (
    cat_df.groupby(["county_name", "sale_year", "month"])["p_logp"]
    .sum()
    .reset_index(name="category_entropy")
)
n_categories = (
    cat_df.groupby(["county_name", "sale_year", "month"])["category_name"]
    .nunique()
    .reset_index(name="n_categories")
)

df = main_df.merge(entropy, on=["county_name", "sale_year", "month"])
df = df.merge(n_categories,  on=["county_name", "sale_year", "month"])
df = df.merge(unemployment_df, on=["sale_year", "month"])  # national monthly rate, same value broadcast to every county that month
df = df.merge(
    homeless_df[["pit_year", "overall_homeless", "overall_chronically_homeless", "unsheltered_homeless"]].rename(columns={"pit_year": "sale_year"}),
    on="sale_year", how="left"  # annual statewide count, same value broadcast to every county-month in that year; left join since homelessness data doesn't yet cover 2026
)
df = df.merge(population_df, on=["county_name", "sale_year"], how="left")  # annual county population, same value broadcast to every month in that year; left join since EL PASO (not a real Iowa county) won't match
df["avg_price_per_bottle"] = df["total_dollars"] / df["total_bottles"]
df["stores_per_10k_capita"] = df["n_stores"] / df["population"] * 10_000  # store density -- normalizes n_stores for county size instead of raw count

print(f"County-month rows: {df.shape[0]}, counties: {df['county_name'].nunique()}")
df.head()
```

```python
feature_cols = [
    "n_stores", "n_vendors", "n_categories",
    "avg_pack", "avg_bottle_volume_ml", "avg_price_per_bottle",
    "category_entropy", "unemployment_rate",
    "overall_homeless", "overall_chronically_homeless", "unsheltered_homeless", "stores_per_10k_capita"
]
X = df[feature_cols].dropna() # X training collumns 

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA()
pcs = pca.fit_transform(X_scaled)

evr = pca.explained_variance_ratio_
cum = np.cumsum(evr)

for i, (e, c) in enumerate(zip(evr, cum), start=1):
    print(f"PC{i}: {e} cumulative: {c}")
```

```python
loadings = pd.DataFrame(
    pca.components_.T * np.sqrt(pca.explained_variance_),
    index=feature_cols,
    columns=[f"PC{i+1}" for i in range(len(feature_cols))],
)
corr = X.corr()

fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(corr.values, cmap="coolwarm", vmin=-1, vmax=1, aspect="auto")

ax.set_xticks(range(corr.shape[1]))
ax.set_xticklabels(corr.columns, rotation=45, ha="right")
ax.set_yticks(range(corr.shape[0]))
ax.set_yticklabels(corr.index)

for i in range(corr.shape[0]):
    for j in range(corr.shape[1]):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center", fontsize=7)

fig.colorbar(im, ax=ax, label="correlation")
ax.set_title("Feature correlation matrix")
plt.tight_layout()
plt.show()



fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(loadings.values, cmap="coolwarm", vmin=-1, vmax=1, aspect="auto")

ax.set_xticks(range(loadings.shape[1]))
ax.set_xticklabels(loadings.columns)
ax.set_yticks(range(loadings.shape[0]))
ax.set_yticklabels(loadings.index)

for i in range(loadings.shape[0]):
    for j in range(loadings.shape[1]):
        ax.text(j, i, f"{loadings.iloc[i, j]:.2f}", ha="center", va="center", fontsize=7)

fig.colorbar(im, ax=ax, label="loading")
ax.set_title("PCA component loadings")
plt.tight_layout()
plt.show()


```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].bar(range(1, len(evr) + 1), evr, color="#4C72B0", label="individual")
axes[0].plot(range(1, len(evr) + 1), cum, color="#C44E52", marker="o", label="cumulative")
axes[0].set_xlabel("Principal component")
axes[0].set_ylabel("Explained variance ratio")
axes[0].set_title("Scree plot")
axes[0].legend()
axes[0].set_ylim(0, 1.05)

axes[1].scatter(pcs[:, 0], pcs[:, 1], s=6, alpha=0.3, color="#4C72B0")
for i, feat in enumerate(feature_cols):
    axes[1].arrow(0, 0, loadings.iloc[i, 0]*3, loadings.iloc[i, 1]*3,
                  color="#C44E52", head_width=0.15, alpha=0.8)
    axes[1].text(loadings.iloc[i, 0]*3.3, loadings.iloc[i, 1]*3.3, feat, fontsize=7, color="#C44E52")
axes[1].set_xlabel("PC1")
axes[1].set_ylabel("PC2")
axes[1].set_title("PC1 vs PC2 (county-months) with feature loadings")
axes[1].axhline(0, color="grey", lw=0.5)
axes[1].axvline(0, color="grey", lw=0.5)

plt.tight_layout()
plt.show()
```

```python
statewide_full = (
    df[df["sale_year"] < 2026]  # 2026 is still in progress -- a partial year would understate its total and distort the trend
    .groupby(["sale_year", "month"])
    .agg(
        total_dollars=("total_dollars", "sum"),
        total_bottles=("total_bottles", "sum"),
        n_stores=("n_stores", "sum"),  # sum of per-county distinct store counts & approximates statewide active stores that month
        n_vendors=("n_vendors", "sum"),  # sum of per-county distinct vendor (brand) counts & approximates statewide active vendors that month
        population=("population", "sum"),  # sum of matched counties' population that year (excludes EL PASO, which has no match)
        unemployment_rate=("unemployment_rate", "mean"),  # same national value for every county that month, so mean == the value
        overall_homeless=("overall_homeless", "mean"),  # annual statewide count, same value for every month in a year, so mean == the value
        overall_chronically_homeless=("overall_chronically_homeless", "mean"),
        unsheltered_homeless=("unsheltered_homeless", "mean"),
    )
    .reset_index()
)
statewide_full["avg_price_per_bottle"] = statewide_full["total_dollars"] / statewide_full["total_bottles"]
statewide_full["stores_per_10k_capita"] = statewide_full["n_stores"] / statewide_full["population"] * 10_000  # recomputed at the aggregate level, not averaged from per-county ratios
statewide_full["time_index"] = (statewide_full["sale_year"] - statewide_full["sale_year"].min()) * 12 + statewide_full["month"]
statewide_full["log_dollars"] = np.log1p(statewide_full["total_dollars"])
sf_month_dummies = pd.get_dummies(statewide_full["month"], prefix="month", drop_first=True)

statewide_full.head()

```

# Forecasting
In earlier models we saw Lasso work the best with the features within our data allowing us to achieve R^2 off 0.44 which by itself is already impressive to achieve. Moreover, our forecast shape was well aligned with our train and test. Now we want to implement a new approach:

1. Train on county-month
2. We should test on different sizes of windows other than a 3 month smoothing. 
3. Switching to a forward CV with a minimum training window floor. 
4. Report a bootstrap prediction interval on the final forecast. 

Note: unemployment doesn't give us any signal. The problem is that its only a point bases of the given month, the unemployment rate is at this level. This doesn't feed any signal to our data. 


```python
panel = q("""
SELECT county_name, sale_year, EXTRACT(month FROM ordered_on) AS month, 
    SUM(sales_dollars) AS total_dollars,
    SUM(sales_bottles) AS total_bottles,
    COUNT(DISTINCT store_no) AS n_stores 
FROM sales 
WHERE county_name IS NOT NULL AND sale_year < 2026
GROUP BY 1, 2, 3 
""")

print(f"panel rows {len(panel)}, shape of data: {panel.shape}, counties: {panel["county_name"].nunique()}")

panel["avg_price_per_bottle"] = panel["total_dollars"] / panel["total_bottles"] # average price for liquor
panel["log_dollars"] = np.log1p(panel["total_dollars"]) # to scale down the cost to logirithmic form
panel = panel.sort_values(["county_name", "sale_year", "month"]).reset_index(drop=True)

# baseline residual, pooled across counties 
month_dummies = pd.get_dummies(panel["month"], prefix="month", drop_first=True)
baseline_cols = ["n_stores", "avg_price_per_bottle"]
baseline_data = pd.concat([panel[baseline_cols + ["log_dollars"]], month_dummies], axis=1).dropna()
X_base = baseline_data[baseline_cols + list(month_dummies.columns)]
y_base = baseline_data["log_dollars"]
baseline_model = LinearRegression().fit(X_base, y_base)
panel.loc[baseline_data.index, "residual"] = y_base - baseline_model.predict(X_base) #Using the residual to be embeded in other model

# county leaks into "lag1" for the next county in row order
g = panel.groupby("county_name")
panel["log_dollars_smooth"] = g["log_dollars"].transform(lambda s: s.rolling(3).mean())
panel["log_dollars_smooth_next"] = g["log_dollars_smooth"].shift(-1)
panel["log_dollars_smooth_lag12"] = g["log_dollars_smooth"].shift(11)
panel["residual_lag1"] = g["residual"].shift(1)
panel["n_stores_lag1"] = g["n_stores"].shift(1)
panel["price_lag1"] = g["avg_price_per_bottle"].shift(1)

forecast_cols = ["log_dollars_smooth_lag12", "residual_lag1", "n_stores_lag1", "price_lag1"]
fc_panel = panel[["county_name", "sale_year", "month"] + forecast_cols + ["log_dollars_smooth_next"]].dropna()
print("fc_panel rows:", len(fc_panel))
```

### Time-Aware Cross Validation with respect to the panel we created
 1. Linear Regression
 2. Ridge
 3. Lasso 
 4. ElasticNet

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error


periods = fc_panel[["sale_year", "month"]].drop_duplicates().sort_values(["sale_year", "month"]).reset_index(drop=True)
tscv = TimeSeriesSplit(n_splits=5)

models = {
    "linearRegression": LinearRegression(),
    "Ridge": make_pipeline(StandardScaler(), Ridge(alpha=1.0)),
    "Lasso": make_pipeline(StandardScaler(), Lasso(alpha=0.01, max_iter=10000)),
    "ElasticNet": make_pipeline(StandardScaler(), ElasticNet(alpha=0.01, l1_ratio=0.5, max_iter=10000)),
}

X_all = fc_panel[forecast_cols]
y_all = fc_panel["log_dollars_smooth_next"]
period_key = list(zip(fc_panel["sale_year"], fc_panel["month"]))

# A Cross Validation (CV) to test and record the CV_r^2 score of a model and be able to compare side by side. 
results = []
for name, model in models.items():
    fold_r2, fold_rmse_log, fold_mae_log, fold_mae_dollars = [], [], [], []
    for train_p_idx, test_p_idx in tscv.split(periods):
        train_periods = set(map(tuple, periods.iloc[train_p_idx][["sale_year", "month"]].values))
        test_periods = set(map(tuple, periods.iloc[test_p_idx][["sale_year", "month"]].values))
        train_mask = np.array([k in train_periods for k in period_key])
        test_mask = np.array([k in test_periods for k in period_key])

        model.fit(X_all[train_mask], y_all[train_mask])
        pred = model.predict(X_all[test_mask])
        actual = y_all[test_mask]

        fold_r2.append(r2_score(actual, pred))
        fold_rmse_log.append(np.sqrt(mean_squared_error(actual, pred)))
        fold_mae_log.append(mean_absolute_error(actual, pred))
        # convert log-smoothed-dollars back to actual dollars for an interpretable error size
        fold_mae_dollars.append(mean_absolute_error(np.expm1(actual), np.expm1(pred)))

    results.append((
        name, np.mean(fold_r2), np.std(fold_r2),
        np.mean(fold_rmse_log), np.mean(fold_mae_log), np.mean(fold_mae_dollars),
    ))

res_df = pd.DataFrame(
    results, columns=["model", "cv_r2_mean", "cv_r2_std", "rmse_log", "mae_log", "mae_dollars"]
).sort_values("cv_r2_mean", ascending=False)
res_df["mae_dollars"] = res_df["mae_dollars"].map(lambda v: f"${v:,.0f}")
print(res_df)
```

```python
naive_scores = []
for train_p_idx, test_p_idx in tscv.split(periods):
    test_periods = set(map(tuple, periods.iloc[test_p_idx][["sale_year", "month"]].values))
    test_mask = np.array([k in test_periods for k in period_key])
    naive_pred = fc_panel.loc[test_mask, "log_dollars_smooth_lag12"]  # "same county, same month last year"
    naive_scores.append(r2_score(y_all[test_mask], naive_pred))
print("Naive (lag12 only, no model) R^2:", np.mean(naive_scores), "+/-", np.std(naive_scores))
```

```python
within_scores = {name: [] for name in models}
for train_p_idx, test_p_idx in tscv.split(periods):
    train_periods = set(map(tuple, periods.iloc[train_p_idx][["sale_year", "month"]].values))
    test_periods = set(map(tuple, periods.iloc[test_p_idx][["sale_year", "month"]].values))
    train_mask = np.array([k in train_periods for k in period_key])
    test_mask = np.array([k in test_periods for k in period_key])

    county_means = y_all[train_mask].groupby(fc_panel.loc[train_mask, "county_name"]).mean()
    overall_mean = y_all[train_mask].mean()
    y_train_dm = y_all[train_mask] - fc_panel.loc[train_mask, "county_name"].map(county_means)
    y_test_dm = y_all[test_mask] - fc_panel.loc[test_mask, "county_name"].map(county_means).fillna(overall_mean)

    for name, model in models.items():
        model.fit(X_all[train_mask], y_train_dm)
        pred = model.predict(X_all[test_mask])
        within_scores[name].append(r2_score(y_test_dm, pred))

for name, scores in within_scores.items():
    print(f"{name}: mean={np.mean(scores):.3f} std={np.std(scores):.3f}")
```

```python
# Fit on ALL available data (final deployed model, not a CV fold)
X_all = fc_panel[forecast_cols]
y_all = fc_panel["log_dollars_smooth_next"]
final_model = make_pipeline(StandardScaler(), Ridge(alpha=1.0))
final_model.fit(X_all, y_all)

# Safe (county, year, month) -> row lookup, instead of position offsets like the
# state-level version used -- panel rows are grouped by county, not pure time order,
# so an idx_last - 12 + h style offset would be wrong here.
lookup = panel.set_index(["county_name", "sale_year", "month"])
last_rows = panel.sort_values(["county_name", "sale_year", "month"]).groupby("county_name").tail(1)

horizons = 3
rows = []
for _, last in last_rows.iterrows():
    county = last["county_name"]
    last_year, last_month = int(last["sale_year"]), int(last["month"])
    for h in range(1, horizons + 1):
        target_month = 1 if (last_month + h - 1) % 12 == 0 else (last_month + h - 1) % 12 + 1
        target_year = last_year + (last_month + h - 1) // 12
        anchor_key = (county, target_year - 1, target_month)  # same month, 12 months back
        if anchor_key not in lookup.index or pd.isna(lookup.loc[anchor_key, "log_dollars_smooth"]):
            continue  # county missing that historical month -- skip rather than guess

        step_X = pd.DataFrame([{
            "log_dollars_smooth_lag12": lookup.loc[anchor_key, "log_dollars_smooth"],
            "residual_lag1": last["residual"],
            "n_stores_lag1": last["n_stores"],
            "price_lag1": last["avg_price_per_bottle"],
        }])[forecast_cols]
        pred_log = final_model.predict(step_X)[0]
        rows.append((county, target_year, target_month, h, np.expm1(pred_log)))

county_forecasts = pd.DataFrame(rows, columns=["county_name", "sale_year", "month", "h", "forecast_dollars"])
statewide_forecast = county_forecasts.groupby(["sale_year", "month"])["forecast_dollars"].sum().reset_index()
statewide_forecast["date"] = pd.to_datetime(dict(year=statewide_forecast["sale_year"], month=statewide_forecast["month"], day=1))
print(statewide_forecast[["date", "forecast_dollars"]])
```
